```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

protocol_1 = "uniswap"
protocol_2 = "pancakeswap"

def get_tvl_data(protocol):
  url = f"https://api.llama.fi/protocol/{protocol}"
  response = requests.get(url)
  data = response.json()

  tvl_data = pd.DataFrame(data["tvl"])
  tvl_data["date"] = pd.to_datetime(tvl_data["date"], unit="s")
  tvl_data["protocol"] = data["name"]
  return tvl_data

tvl1 = get_tvl_data(protocol_1)
tvl2 = get_tvl_data(protocol_2)

merged = pd.concat([tvl1, tvl2])
merged = merged.sort_values("date")

plt.figure(figsize=(10,6))
for protocol, df in merged.groupby("protocol"):
  plt.plot(df["date"], df["totalLiquidityUSD"], label=protocol)

plt.title("Comparativa de TVL entre Protocolos DEFI")
plt.xlabel("Fecha")
plt.ylabel("TVL")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
